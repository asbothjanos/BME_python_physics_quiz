```python
# Add custom CSS so long text wraps nicely across lines:

# Question database strictly adhering to course constraints
quiz_data = [
    {
        "question": "Q1: Which Python loop correctly updates a particle's position x using the Euler algorithm x = x + v * dt over 100 time steps?",
        "options": [
            "A) for i in range(100):\n    x = x + v * dt",
            "B) while i < 100:\n    x = x + v * dt",
            "C) for x in range(100):\n    x = x + v * dt",
            "D) loop 100 times:\n    x = x + v * dt",
        ],
        "correct": "A",
        "explanation": "range(100) loops exactly 100 times, cleanly updating the scalar position variable x at each step.",
    },
    {
        "question": "Q2: To create an array of 50 equally spaced time values between t = 0 and t = 10 seconds in NumPy, which command should be used?",
        "options": [
            "A) np.arange(0, 10, 50)",
            "B) np.linspace(0, 10, 50)",
            "C) np.linspace(0, 50, 10)",
            "D) np.array(0, 10, 50)",
        ],
        "correct": "B",
        "explanation": "np.linspace(start, stop, num) generates num evenly spaced points including start and stop.",
    },
    {
        "question": "Q3: Given a 1D NumPy array pos storing position values at each time step, which code calculates velocity using forward finite differences v_i = (x_{i+1} - x_i) / dt without explicit Python loops?",
        "options": [
            "A) v = (pos[1:] - pos[:-1]) / dt",
            "B) v = (pos[1:] - pos[0:]) / dt",
            "C) v = (pos[:-1] - pos[1:]) / dt",
            "D) v = (pos[1:50] - pos[0:50]) / dt",
        ],
        "correct": "A",
        "explanation": "pos[1:] shifts elements forward by one index relative to pos[:-1], allowing vector subtraction of adjacent elements.",
    },
    {
        "question": "Q4: Given a NumPy array pos = np.array([10, 20, 30, 40, 50]), what does the slice pos[1:4] return?",
        "options": [
            "A) np.array([10, 20, 30])",
            "B) np.array([20, 30, 40])",
            "C) np.array([20, 30, 40, 50])",
            "D) np.array([10, 20, 30, 40])",
        ],
        "correct": "B",
        "explanation": "Slicing pos[start:stop] starts at index 1 (20) and stops before index 4 (50), yielding elements at indices 1, 2, and 3.",
    },
    {
        "question": "Q5: Which line of code correctly computes potential energy E = 0.5 * k * x**2 for a NumPy array of positions x and scalar spring constant k?",
        "options": [
            "A) ep = 0.5 * k * x2",
            "B) ep = 0.5 * k * x**2",
            "C) ep = 1/2 * k * x^2",
            "D) ep = 0.5 * k * x * 2",
        ],
        "correct": "B",
        "explanation": "Python requires ** for exponentiation (x**2). Operations on array x automatically vectorize.",
    },
    {
        "question": "Q6: You have a time array t and position array x. Which Matplotlib sequence generates a correctly labeled line plot of position against time?",
        "options": [
            "A) plt.plot(t, x)\nplt.xlabel('Time (s)')\nplt.ylabel('Position (m)')\nplt.show()",
            "B) plt.line(t, x)\nplt.setlabel('Time (s)', 'Position (m)')",
            "C) plt.plot(x, t)\nplt.xlabel('Time (s)')\nplt.ylabel('Position (m)')\nplt.show()",
            "D) plt.draw(t, x)\nplt.labelx('Time (s)')\nplt.labely('Position (m)')",
        ],
        "correct": "A",
        "explanation": "plt.plot(x_data, y_data) puts independent variable t on the horizontal axis and position x on the vertical axis.",
    },
    {
        "question": "Q7: Consider the code: a = np.array([1, 2, 3]); b = a; b[0] = 99. What is the value of a[0] after execution?",
        "options": [
            "A) 1",
            "B) 99",
            "C) 0",
            "D) Raises an Error",
        ],
        "correct": "B",
        "explanation": "Assigning array variable b = a creates a reference to the same memory space. Modifying b updates a.",
    },
    {
        "question": "Q8: A student wants to compute analytical trajectory values y(t) = v0 * t - 0.5 * g * t**2. Which option uses vectorization correctly for a NumPy array t?",
        "options": [
            "A) y = v0 * t - 0.5 * g * t**2",
            "B) y = v0 * t - 1/2 * g * t2",
            "C) y = v0 * t - 0.5 * g * t^2",
            "D) y = v0 * t - 0.5 * g * t * 2",
        ],
        "correct": "A",
        "explanation": "NumPy arrays multiply and exponentiate element-by-element using standard arithmetic operators.",
    },
    {
        "question": "Q9: Which Python function definition correctly computes kinetic energy K = 0.5 * m * v**2 given mass m and velocity v?",
        "options": [
            "A) def getke(m, v):\n    return 0.5 * m * v**2",
            "B) def getke(m, v):\n    ke = 0.5 * m * v2\n    print(ke)",
            "C) function getke(m, v):\n    return 0.5 * m * v^2",
            "D) def getke(m, v):\n    return 1/2 * m * v2",
        ],
        "correct": "A",
        "explanation": "Python function definitions start with def, power requires **, and values must be passed back using return.",
    },
    {
        "question": "Q10: How do you plot numerical trajectory points as discrete red dots and the analytical trajectory as a blue solid line on the same Matplotlib plot?",
        "options": [
            "A) plt.plot(tnum, ynum, 'ro')\nplt.plot(tana, yana, 'b-')",
            "B) plt.plot(tnum, ynum, 'red-dots')\nplt.plot(tana, yana, 'blue-line')",
            "C) plt.plot(tnum, ynum, style='dot')\nplt.plot(tana, yana, style='line')",
            "D) plt.scatter_line(tnum, ynum, tana, yana)",
        ],
        "correct": "A",
        "explanation": "'ro' formats points as red (r) circles (o) without connecting lines, while 'b-' formats blue (b) continuous lines (-).",
    },
]
import ipywidgets as widgets
from IPython.display import display, clear_output, Markdown

def run_quiz(quiz_data):
    for item in quiz_data:
        # Render the question and options using Markdown for perfect wrapping & code styling
        md_text = f"### {item['question']}\n\n"
        for opt in item["options"]:
            md_text += f"{opt}\n\n"

        display(Markdown(md_text))

        # Compact choice selector
        options_widget = widgets.RadioButtons(
            options=['A', 'B', 'C', 'D'],
            description='Select:',
            disabled=False,
            layout=widgets.Layout(width='auto')
        )

        submit_btn = widgets.Button(description="Check Answer", button_style='primary')
        out = widgets.Output()

        def check_answer(b, item=item, opts=options_widget, out=out):
            with out:
                clear_output()
                selected = opts.value
                if selected == item["correct"]:
                    display(Markdown(f"**Correct!** {item['explanation']}"))
                else:
                    display(Markdown(f"**Incorrect.** The correct answer is **{item['correct']}**. {item['explanation']}"))

        submit_btn.on_click(check_answer)
        display(options_widget, submit_btn, out)
        display(Markdown("---"))

run_quiz(quiz_data)
```
